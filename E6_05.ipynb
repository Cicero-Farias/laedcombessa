```python
class Node:
    def __init__(self, value, index):
        self.value = value
        self.index = index
        self.prev = None
        self.next = None
```

```python
class SparseVector:

    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, value, index):

        novo = Node(value, index)

        if self.head is None:
            self.head = novo
            self.tail = novo
            return

        novo.prev = self.tail
        self.tail.next = novo
        self.tail = novo

    def display(self):

        atual = self.head

        while atual:
            print(f"({atual.index}, {atual.value})", end=" <-> ")
            atual = atual.next

        print("None")
```

```python
def busca_por_indice(lista, k):

    atual = lista.head

    while atual is not None:

        if atual.index == k:
            return atual.value

        atual = atual.next

    return 0
```

```python
def busca_por_valor(lista, x):

    atual = lista.head

    while atual is not None:

        if atual.value == x:
            return atual.index

        atual = atual.next

    return -1
```

```python
def atualizar(lista, x, k):

    atual = lista.head

    while atual is not None:

        if atual.index == k:

            # Remove o nó se o novo valor for zero
            if x == 0:

                if atual.prev is not None:
                    atual.prev.next = atual.next
                else:
                    lista.head = atual.next

                if atual.next is not None:
                    atual.next.prev = atual.prev
                else:
                    lista.tail = atual.prev

            else:
                atual.value = x

            return

        atual = atual.next

    # Se o índice não existir e o valor for diferente de zero,
    # insere um novo nó no final.
    if x != 0:
        lista.append(x, k)
```

```python
vetor = SparseVector()

vetor.append(4, 3)
vetor.append(5, 7)
vetor.append(10, 9)
vetor.append(1, 12)
vetor.append(9, 17)

print("Vetor esparso:")
vetor.display()

print("\nBusca por índice 9:")
print(busca_por_indice(vetor, 9))

print("\nBusca por índice 5:")
print(busca_por_indice(vetor, 5))

print("\nBusca por valor 10:")
print(busca_por_valor(vetor, 10))

print("\nBusca por valor 100:")
print(busca_por_valor(vetor, 100))

print("\nAtualizando posição 9 para 20...")
atualizar(vetor, 20, 9)
vetor.display()

```
