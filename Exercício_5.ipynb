```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:

    def __init__(self):
        self.head = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            return

        atual = self.head
        while atual.next:
            atual = atual.next

        atual.next = novo

    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("None")
```

```python
def intercalar_listas(lista1, lista2):

    resultado = LinkedList()

    p1 = lista1.head
    p2 = lista2.head

    while p1 is not None and p2 is not None:

        if p1.value <= p2.value:
            resultado.append(p1.value)
            p1 = p1.next
        else:
            resultado.append(p2.value)
            p2 = p2.next

    while p1 is not None:
        resultado.append(p1.value)
        p1 = p1.next

    while p2 is not None:
        resultado.append(p2.value)
        p2 = p2.next

    return resultado
```

```python
lista1 = LinkedList()
lista2 = LinkedList()

for valor in [3, 6, 7, 10, 13]:
    lista1.append(valor)

for valor in [2, 4, 9, 11, 12]:
    lista2.append(valor)

print("Lista 1:")
lista1.display()

print("Lista 2:")
lista2.display()

resultado = intercalar_listas(lista1, lista2)

print("\nLista intercalada:")
resultado.display()
```
