```python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None
```

```python
class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, value):

        novo = Node(value)

        if self.head is None:
            self.head = novo
            self.tail = novo
            return

        novo.prev = self.tail
        self.tail.next = novo
        self.tail = novo

    def display(self):

        atual = self.head

        while atual:
            print(atual.value, end=" <-> ")
            atual = atual.next

        print("None")
```

```python
def intercalar(lista1, lista2):

    resultado = DoublyLinkedList()

    p1 = lista1.head
    p2 = lista2.head

    while p1 is not None and p2 is not None:

        if p1.value <= p2.value:
            resultado.append(p1.value)
            p1 = p1.next
        else:
            resultado.append(p2.value)
            p2 = p2.next

    while p1 is not None:
        resultado.append(p1.value)
        p1 = p1.next

    while p2 is not None:
        resultado.append(p2.value)
        p2 = p2.next

    return resultado
```

```python
lista1 = DoublyLinkedList()
lista2 = DoublyLinkedList()

for valor in [2, 5, 8, 11]:
    lista1.append(valor)

for valor in [1, 3, 7, 9, 12]:
    lista2.append(valor)

print("Lista 1:")
lista1.display()

print("Lista 2:")
lista2.display()

resultado = intercalar(lista1, lista2)

print("\nLista intercalada:")
resultado.display()
```

Tempo: O(n + m), onde n é o número de elementos da primeira lista e m o da segunda, pois cada elemento é visitado exatamente uma vez.
