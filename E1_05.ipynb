```python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None
```

```python
class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            self.tail = novo
            return

        novo.prev = self.tail
        self.tail.next = novo
        self.tail = novo

    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" <-> ")
            atual = atual.next

        print("None")
```

```python
def elemento_central(lista):

    if lista.head is None:
        return None

    lento = lista.head
    rapido = lista.head

    while rapido.next is not None and rapido.next.next is not None:
        lento = lento.next
        rapido = rapido.next.next

    return lento.value
```

```python
lista = DoublyLinkedList()

for valor in [3, 9, 5, 2, 8]:
    lista.append(valor)

print("Lista:")
lista.display()

print("Elemento central:", elemento_central(lista))
```

```python
lista2 = DoublyLinkedList()

for valor in [1, 3, 5, 7, 9, 11]:
    lista2.append(valor)

print("Lista:")
lista2.display()

print("Elemento central:", elemento_central(lista2))
```

Tempo: O(n), pois a lista é percorrida apenas uma vez.
