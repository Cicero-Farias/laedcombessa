```python
def intersecao(U, V):
    if not U or not V:
        return []
    if len(U) <= len(V):
        menor, maior = U, V
    else:
        menor, maior = V, U

    conjunto_menor = set(menor)
    inter = set()
    for x in maior:
        if x in conjunto_menor:
            inter.add(x)

    return sorted(inter)

U = [9, 2, 7, 7, 2, 2, 1, 7, 7, 9]
V = [2, 15, 19, 12, 33, 9, 17, 41, 54, 8]
print(intersecao(U, V))
```
