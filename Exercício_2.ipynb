```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:

    def __init__(self):
        self.head = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            return

        atual = self.head
        while atual.next:
            atual = atual.next

        atual.next = novo

    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("None")
```

```python
def separar_pares_impares(lista):

    lista_pares = LinkedList()
    lista_impares = LinkedList()

    atual = lista.head

    while atual:

        if atual.value % 2 == 0:
            lista_pares.append(atual.value)
        else:
            lista_impares.append(atual.value)

        atual = atual.next

    return lista_impares, lista_pares
```

```python
lista = LinkedList()

for valor in [2, 8, 5, 10, 7]:
    lista.append(valor)

print("Lista original:")
lista.display()

impares, pares = separar_pares_impares(lista)

print("\nLista de ímpares:")
impares.display()

print("Lista de pares:")
pares.display()
```
