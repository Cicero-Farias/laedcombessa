```python
def mediana(lista):
    n = len(lista)

    if n % 2 == 0:
        return (lista[n // 2] + lista[n // 2 - 1]) / 2
    else:
        return lista[n // 2]


def mediana_duas_listas(U, V):
    n = len(U)

    if n == 1:
        return (U[0] + V[0]) / 2

    if n == 2:
        return (max(U[0], V[0]) + min(U[1], V[1])) / 2

    m1 = mediana(U)
    m2 = mediana(V)

    if m1 == m2:
        return m1

    if m1 < m2:
        if n % 2 == 0:
            return mediana_duas_listas(U[n//2 - 1:], V[:n//2 + 1])
        else:
            return mediana_duas_listas(U[n//2:], V[:n//2 + 1])

    else:
        if n % 2 == 0:
            return mediana_duas_listas(U[:n//2 + 1], V[n//2 - 1:])
        else:
            return mediana_duas_listas(U[:n//2 + 1], V[n//2:])


U = [1, 3, 5, 7, 9]
V = [2, 4, 6, 8, 10]

resultado = mediana_duas_listas(U, V)

print("Mediana:", resultado)
```
