```python
def busca_aproximada(V, k):
    if not V:
        return None, False

    melhor = V[0]
    for x in V:
        if x == k:
            return k, True
        if abs(x - k) < abs(melhor - k):
            melhor = x
        elif abs(x - k) == abs(melhor - k) and x < melhor:
            melhor = x
    return melhor, False

V = [9, 42, 21, 14, 25, 3, 19, 33, 45, 6]
k = 31
resultado, achou = busca_aproximada(V, k)
if achou:
    print(f"{resultado} (está lá)")
else:
    print(f"Não está lá, mas eu encontrei o {resultado}")
```
