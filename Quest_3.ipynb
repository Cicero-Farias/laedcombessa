```python
def mais_proximo_da_media(v):
    media = sum(v) / len(v)

    mais_proximo = v[0]
    menor_distancia = abs(v[0] - media)

    for x in v:
        distancia = abs(x - media)

        if distancia < menor_distancia:
            menor_distancia = distancia
            mais_proximo = x

    return mais_proximo, media

v = [5, 3, 1, 10, 2, 13, 9, 12, 4, 7]

resultado, media = mais_proximo_da_media(v)

print("Elemento:", resultado)
print("Média:", media)
```
