```python
def achar_impar_com_vezes_impar(V):
    cont = {}
    for x in V:
        cont[x] = cont.get(x, 0) + 1

    for valor, freq in cont.items():
        if valor % 2 != 0 and freq % 2 != 0:
            return valor
    return None

V = [9, 2, 7, 7, 2, 2, 1, 7, 7, 9]

def lista_impares_com_freq_impar(V):
    cont = {}
    for x in V:
        cont[x] = cont.get(x, 0) + 1
    resultado = []
    for valor, freq in cont.items():
        if valor % 2 != 0 and freq % 2 != 0:
            resultado.append(valor)
    return resultado

print(lista_impares_com_freq_impar(V))
```
