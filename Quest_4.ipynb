```python
def elemento_isolado(v):
    s = set(v)

    for x in v:
        if (x - 1 not in s) and (x + 1 not in s):
            return True, x

    return False, None

v = [17, 2, 8, 1, 7, 13, 9, 12, 4, 16]

print(elemento_isolado(v))
```
