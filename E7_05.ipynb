```python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None
```

```python
class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, value):

        novo = Node(value)

        if self.head is None:
            self.head = novo
            self.tail = novo
            return

        novo.prev = self.tail
        self.tail.next = novo
        self.tail = novo

    def display(self):

        atual = self.head

        while atual:
            print(atual.value, end=" <-> ")
            atual = atual.next

        print("None")
```

```python
def eh_palindromo(lista):

    esquerda = lista.head
    direita = lista.tail

    while esquerda is not None and direita is not None:

        if esquerda.value != direita.value:
            return False

        # Os ponteiros se encontraram (tamanho ímpar)
        if esquerda == direita:
            break

        # Os ponteiros se cruzaram (tamanho par)
        if esquerda.next == direita:
            break

        esquerda = esquerda.next
        direita = direita.prev

    return True
```

```python
lista = DoublyLinkedList()

for valor in [1, 2, 3, 2, 1]:
    lista.append(valor)

lista.display()

if eh_palindromo(lista):
    print("A lista é um palíndromo.")
else:
    print("A lista não é um palíndromo.")
```

```python
lista = DoublyLinkedList()

for valor in [4, 8, 8, 4]:
    lista.append(valor)

lista.display()

if eh_palindromo(lista):
    print("A lista é um palíndromo.")
else:
    print("A lista não é um palíndromo.")
```

```python
lista = DoublyLinkedList()

for valor in [1, 2, 3, 4]:
    lista.append(valor)

lista.display()

if eh_palindromo(lista):
    print("A lista é um palíndromo.")
else:
    print("A lista não é um palíndromo.")
```

Tempo: O(n), pois cada comparação envolve um par de elementos e os ponteiros percorrem a lista até o centro.
