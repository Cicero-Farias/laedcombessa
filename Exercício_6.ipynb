```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:

    def __init__(self):
        self.head = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            return

        atual = self.head
        while atual.next:
            atual = atual.next

        atual.next = novo

    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("None")
```

```python
def particionar(lista, k):

    menores = LinkedList()
    maiores = LinkedList()

    atual = lista.head

    while atual:

        if atual.value <= k:
            menores.append(atual.value)
        else:
            maiores.append(atual.value)

        atual = atual.next

    # Se não houver elementos <= k
    if menores.head is None:
        lista.head = maiores.head
        return

    # Encontra o final da lista de menores
    atual = menores.head
    while atual.next:
        atual = atual.next

    # Liga as duas listas
    atual.next = maiores.head

    # Atualiza a lista original
    lista.head = menores.head
```

```python
lista = LinkedList()

for valor in [9, 2, 5, 6, 1]:
    lista.append(valor)

print("Lista original:")
lista.display()

k = 5

particionar(lista, k)

print(f"\nLista após a partição (k = {k}):")
lista.display()
```
