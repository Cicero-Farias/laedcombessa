```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:

    def __init__(self):
        self.head = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            return

        atual = self.head
        while atual.next:
            atual = atual.next

        atual.next = novo

    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("None")
```

```python
def inverter_lista(lista):

    anterior = None
    atual = lista.head

    while atual:

        proximo = atual.next
        atual.next = anterior
        anterior = atual
        atual = proximo

    lista.head = anterior
```

```python
lista = LinkedList()

for valor in [3, 2, 5, 9, 4]:
    lista.append(valor)

print("Antes:")
lista.display()

inverter_lista(lista)

print("\nDepois:")
lista.display()
```
