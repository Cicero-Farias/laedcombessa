```python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None
```

```python
class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, value):

        novo = Node(value)

        if self.head is None:
            self.head = novo
            self.tail = novo
            return

        novo.prev = self.tail
        self.tail.next = novo
        self.tail = novo

    def display(self):

        atual = self.head

        while atual:
            print(atual.value, end=" <-> ")
            atual = atual.next

        print("None")
```

```python
def remover_repetidos(lista):

    vistos = set()
    atual = lista.head

    while atual is not None:

        proximo = atual.next

        if atual.value in vistos:

            # Atualiza o ponteiro do nó anterior
            if atual.prev is not None:
                atual.prev.next = atual.next
            else:
                lista.head = atual.next

            # Atualiza o ponteiro do próximo nó
            if atual.next is not None:
                atual.next.prev = atual.prev
            else:
                lista.tail = atual.prev

        else:
            vistos.add(atual.value)

        atual = proximo
```

```python
lista = DoublyLinkedList()

for valor in [3, 5, 3, 8, 5, 2, 8, 10]:
    lista.append(valor)

print("Lista original:")
lista.display()

remover_repetidos(lista)

print("\nLista sem repetidos:")
lista.display()
```

Tempo: O(n), pois cada nó é visitado apenas uma vez e as operações no conjunto (set) possuem custo médio O(1).
