a) O Radix Sort em base 10 (variante LSD, que processa do dígito menos significativo para o mais significativo) depende da estabilidade da etapa de coleta. A distribuição coloca cada elemento na fila correspondente ao dígito atual, e a coleta deve retirar mantendo a ordem relativa entre elementos que caem na mesma fila. Se, no lugar de filas, forem usadas pilhas na coleta, a ordem relativa desses elementos com o mesmo dígito pode ser invertida. Isso quebra o efeito acumulado do processamento de dígitos: os dígitos mais significativos passam a “misturar” elementos que já haviam sido ordenados corretamente pelos dígitos menos significativos. Um contra-exemplo simples é a entrada 11, 21. No passo das unidades, ambos têm unidade 1. Com filas, a coleta devolve [11, 21], preservando a ordem; com pilhas (LIFO), a coleta devolve [21, 11], invertendo a ordem dos elementos que compartilham o mesmo dígito. Depois, ao seguir para o próximo dígito, essa inversão pode impedir que o resultado final seja a ordenação correta, porque o Radix LSD só funciona se cada passagem for estável.

b) Para traçar o Radix Sort da sequência 481, 329, 143, 612, 937, 480, 256, considere base 10 e três passagens: unidades, dezenas e centenas. Na primeira passagem (unidades), os dígitos das unidades são: 481→1, 329→9, 143→3, 612→2, 937→7, 480→0, 256→6. Assim, as filas ficam F0=[480], F1=[481], F2=[612], F3=[143], F6=[256], F7=[937], F9=[329], e as demais vazias. Coletando em ordem F0 até F9, obtém-se a sequência após a passagem 1: [480, 481, 612, 143, 256, 937, 329]. Na segunda passagem (dezenas), para essa nova sequência os dígitos das dezenas são: 480→8, 481→8, 612→1, 143→4, 256→5, 937→3, 329→2. Logo, as filas ficam F1=[612], F2=[329], F3=[937], F4=[143], F5=[256] e F8=[480, 481] (mantendo 480 antes de 481, pois a estabilidade preserva a ordem de entrada). Coletando F0..F9, a sequência fica [612, 329, 937, 143, 256, 480, 481]. Na terceira passagem (centenas), os dígitos das centenas são: 612→6, 329→3, 937→9, 143→1, 256→2, 480→4, 481→4. Então F1=[143], F2=[256], F3=[329], F4=[480, 481], F6=[612] e F9=[937]. Coletando, a sequência final é [143, 256, 329, 480, 481, 612, 937].

c) Para strings de comprimento fixo ℓ sobre um alfabeto de tamanho σ, o Radix Sort (com uma passagem por posição, da última para a primeira no modo LSD) faz ℓ passagens. Em cada passagem, distribui os n elementos em σ “baldes” e depois coleta, o que custa Θ(n + σ) por passagem. Portanto o tempo total é Θ(ℓ(n + σ)). Se σ for constante, isso vira Θ(ℓn). Em espaço, é necessário armazenamento auxiliar para a distribuição/coleta, além das estruturas dos baldes, resultando em Θ(n + σ). Comparando com Merge Sort, para ordenar strings tipicamente conta-se Θ(n log n) comparações, mas comparar duas strings pode custar até Θ(ℓ) no pior caso, porque pode ser necessário verificar muitos caracteres até decidir. Assim, um custo comum para Merge Sort nesse cenário é Θ(ℓ n log n).
