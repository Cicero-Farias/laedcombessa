```python
class Node:

    def __init__(self, value):
        self.value = value
        self.next = None


class Stack:

    def __init__(self):
        self.top = None

    def push(self, value):

        novo = Node(value)
        novo.next = self.top
        self.top = novo

    def pop(self):

        if self.top is None:
            return None

        valor = self.top.value
        self.top = self.top.next

        return valor

    def display(self):

        atual = self.top

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("/")
```

```python
class MultiNode:

    def __init__(self, value):

        self.value = value
        self.next = None
        self.subPilha = Stack()
```

```python
class MultiLevelStack:

    def __init__(self):
        self.top = None

    def push(self, value):

        novo = MultiNode(value)

        novo.next = self.top
        self.top = novo

    def pop(self):

        if self.top is None:
            return None

        valor = self.top.value
        self.top = self.top.next

        return valor

    def display(self):

        atual = self.top

        while atual:

            print(atual.value, end="")

            if atual.subPilha.top is not None:

                print("  { ", end="")
                atual.subPilha.display()
                print("}", end="")

            print()

            atual = atual.next
```

```python
P = MultiLevelStack()

P.push(10)
P.push(20)
P.push(30)

print("Pilha principal:")
P.display()

print("\nInserindo elementos na subpilha do topo...")

P.top.subPilha.push(100)
P.top.subPilha.push(200)
P.top.subPilha.push(300)

P.display()

print("\nPop da pilha principal:", P.pop())

print("\nEstado final:")
P.display()
```

Push(P, x):

cria um novo nó;
faz o novo nó apontar para o antigo topo;
atualiza o topo da pilha.

Pop(P):

verifica se a pilha está vazia;
guarda o elemento do topo;
atualiza o topo para o próximo nó;
retorna o elemento removido.

A instrução Push(P.top.subPilha, y) insere o elemento y na subpilha associada ao elemento que está atualmente no topo da pilha principal. Dessa forma, a pilha principal permanece inalterada e apenas a subpilha pertencente ao nó do topo recebe um novo elemento. Isso permite que cada elemento da pilha principal mantenha sua própria pilha de dados independente das demais.

As operações Push e Pop da pilha principal executam apenas a criação ou remoção do nó localizado no topo da estrutura, sem percorrer seus elementos, possuindo complexidade O(1). Da mesma forma, a operação Push(P.top.subPilha, y) atua apenas sobre a subpilha associada ao nó do topo, inserindo um novo elemento no início dessa subpilha, também em tempo O(1). Assim, todas as operações solicitadas no exercício possuem tempo constante e utilizam apenas espaço auxiliar constante.
