```python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None
```

```python
class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, value):

        novo = Node(value)

        if self.head is None:
            self.head = novo
            self.tail = novo
            return

        novo.prev = self.tail
        self.tail.next = novo
        self.tail = novo

    def display(self):

        atual = self.head

        while atual:
            print(atual.value, end=" <-> ")
            atual = atual.next

        print("None")
```

```python
def lista_de_listas(lista, k):

    if k <= 0:
        raise ValueError("k deve ser maior que zero.")

    vetor = []

    atual = lista.head

    while atual is not None:

        sublista = DoublyLinkedList()

        contador = 0

        while atual is not None and contador < k:
            sublista.append(atual.value)
            atual = atual.next
            contador += 1

        vetor.append(sublista)

    return vetor
```

```python
def mostrar_lista_de_listas(vetor):

    for i, sublista in enumerate(vetor):

        print(f"Sublista {i+1}: ", end="")
        sublista.display()
```

```python
lista = DoublyLinkedList()

for valor in [1, 3, 5, 7, 10, 13, 15, 18, 21, 24, 27]:
    lista.append(valor)

print("Lista original:")
lista.display()

k = 3

L = lista_de_listas(lista, k)

print(f"\nLista de listas (k = {k}):")
mostrar_lista_de_listas(L)
```

Se a lista possui n elementos, cada elemento é visitado exatamente uma vez e a criação das sublistas ocorre durante essa única passagem. Portanto, o tempo é O(n).
