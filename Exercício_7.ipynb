```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:

    def __init__(self):
        self.head = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            return

        atual = self.head

        while atual.next:
            atual = atual.next

        atual.next = novo

    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("None")
```

```python
def remover_copias(lista, k):

    # Remove ocorrências no início da lista
    while lista.head is not None and lista.head.value == k:
        lista.head = lista.head.next

    # Se a lista ficou vazia
    if lista.head is None:
        return

    anterior = lista.head
    atual = lista.head.next

    while atual is not None:

        if atual.value == k:
            anterior.next = atual.next
        else:
            anterior = atual

        atual = atual.next
```

```python
lista = LinkedList()

for valor in [1, 3, 3, 2, 3, 2]:
    lista.append(valor)

print("Lista original:")
lista.display()

k = 3

remover_copias(lista, k)

print(f"\nLista após remover todas as ocorrências de {k}:")
lista.display()
```
