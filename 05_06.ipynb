```python
class Stack:

    def __init__(self):
        self.items = []

    def push(self, value):
        self.items.append(value)

    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0
```

```python
def balanceada(expressao):

    pilha = Stack()

    pares = {
        ')': '(',
        ']': '[',
        '}': '{'
    }

    for caractere in expressao:

        if caractere in "([{":
            pilha.push(caractere)

        elif caractere in ")]}":

            if pilha.is_empty():
                return False

            topo = pilha.pop()

            if topo != pares[caractere]:
                return False

    return pilha.is_empty()
```

```python
expressoes = [

    "(A+B)*C",

    "((A+B)*C)",

    "{[A+(B*C)]}",

    "(A+B]",

    "((A+B)",

    "{[(])}",

    "A+B*C"

]

for e in expressoes:

    print(e)

    if balanceada(e):
        print("Balanceada\n")
    else:
        print("Não balanceada\n")
```

A pilha garante que o último símbolo de abertura encontrado seja o primeiro a ser fechado, obedecendo ao comportamento LIFO (Last In, First Out). Dessa forma, cada símbolo de fechamento é comparado com o topo da pilha, assegurando que a ordem de abertura e fechamento esteja correta. Se ocorrer um fechamento sem abertura correspondente, um fechamento incompatível ou permanecerem símbolos de abertura na pilha ao final da leitura, a expressão não está corretamente balanceada.

O algoritmo percorre a expressão apenas uma vez. Para cada caractere, realiza operações de empilhar, desempilhar ou consultar o topo da pilha, todas com custo constante O(1). Assim, considerando uma expressão com n caracteres, o tempo total de execução é O(n). O espaço auxiliar também é O(n) no pior caso, quando todos os caracteres iniciais da expressão são símbolos de abertura.
