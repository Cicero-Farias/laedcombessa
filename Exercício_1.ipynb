```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:

    def __init__(self):
        self.head = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            return

        atual = self.head
        while atual.next:
            atual = atual.next

        atual.next = novo


    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("None")
```

```python
def mover_maior_para_fim(lista):

    # lista vazia ou com um único elemento
    if lista.head is None or lista.head.next is None:
        return

    maior = lista.head
    anterior_maior = None

    anterior = None
    atual = lista.head

    ultimo = None

    while atual:

        if atual.value > maior.value:
            maior = atual
            anterior_maior = anterior

        ultimo = atual
        anterior = atual
        atual = atual.next

    # já está no fim
    if maior == ultimo:
        return

    # remove o maior da posição atual
    if anterior_maior is None:
        lista.head = maior.next
    else:
        anterior_maior.next = maior.next

    # coloca no final
    ultimo.next = maior
    maior.next = None
```

```python
lista = LinkedList()

for valor in [5, 8, 13, 2, 10]:
    lista.append(valor)

print("Antes:")
lista.display()

mover_maior_para_fim(lista)

print("\nDepois:")
lista.display()
```
