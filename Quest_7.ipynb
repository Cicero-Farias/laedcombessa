```python
def repetidos_proximos(v, k):
    ultima_posicao = {}

    for i in range(len(v)):
        x = v[i]

        if x in ultima_posicao:
            distancia = i - ultima_posicao[x]

            if distancia <= k:
                return True, x, ultima_posicao[x], i

        ultima_posicao[x] = i

    return False, None

v = [2, 1, 9, 7, 6, 3, 9, 4, 2, 6, 1, 3]

print(repetidos_proximos(v, 4))
```
