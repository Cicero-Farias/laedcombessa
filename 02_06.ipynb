```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class Queue:

    def __init__(self):
        self.inicio = None
        self.fim = None

    # Inserção no fim
    def enqueue(self, value):

        novo = Node(value)

        if self.inicio is None:
            self.inicio = novo
            self.fim = novo
            return

        self.fim.next = novo
        self.fim = novo

    # Remoção no início
    def dequeue(self):

        if self.inicio is None:
            print("Fila vazia.")
            return None

        removido = self.inicio
        self.inicio = self.inicio.next

        # Se a fila ficou vazia
        if self.inicio is None:
            self.fim = None

        return removido.value

    def display(self):

        atual = self.inicio

        print("inicio -> ", end="")

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("/")

        if self.fim:
            print("fim =", self.fim.value)
        else:
            print("fim = None")
```

```python
fila = Queue()

fila.enqueue(8)
fila.enqueue(15)
fila.enqueue(23)

print("Fila inicial:")
fila.display()
```

```python
fila.enqueue(7)
fila.enqueue(11)

print("Após Enqueue(7) e Enqueue(11):")
fila.display()
```

```python
r1 = fila.dequeue()

print("Primeiro Dequeue:")
print("Valor removido:", r1)
fila.display()

r2 = fila.dequeue()

print("\nSegundo Dequeue:")
print("Valor removido:", r2)
fila.display()
```

Quando a última remoção esvazia a fila, os ponteiros inicio e fim devem ser atualizados para None, indicando que a estrutura não possui mais elementos. Se apenas inicio for zerado e fim continuar apontando para o último nó removido, a fila ficará em um estado inconsistente, pois fim fará referência a um nó que já não pertence mais à estrutura. Isso pode provocar erros em futuras operações de inserção ou remoção, além de manter uma referência inválida ao último elemento.
