```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:

    def __init__(self):
        self.head = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            return

        atual = self.head
        while atual.next:
            atual = atual.next

        atual.next = novo

    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("None")
```

```python
def duplicar_impares(lista):

    atual = lista.head

    while atual:

        if atual.value % 2 != 0:
            copia = Node(atual.value)

            copia.next = atual.next
            atual.next = copia

            # Avança para o próximo elemento original
            atual = copia.next
        else:
            atual = atual.next
```

```python
lista = LinkedList()

for valor in [2, 7, 6, 3]:
    lista.append(valor)

print("Antes:")
lista.display()

duplicar_impares(lista)

print("\nDepois:")
lista.display()
```
