```python
class Stack:

    def __init__(self):
        self.items = []

    def push(self, value):
        self.items.append(value)

    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def display(self):
        print(self.items)
```

```python
class QueueWithTwoStacks:

    def __init__(self):
        self.P1 = Stack()   # entrada
        self.P2 = Stack()   # saída

    def enqueue(self, value):
        self.P1.push(value)

    def dequeue(self):

        if self.P2.is_empty():

            while not self.P1.is_empty():
                self.P2.push(self.P1.pop())

        if self.P2.is_empty():
            return None

        return self.P2.pop()

    def display(self):

        print("P1 (entrada):", self.P1.items)
        print("P2 (saída)  :", self.P2.items)
```

```python
fila = QueueWithTwoStacks()

for x in [10, 20, 30, 40]:
    fila.enqueue(x)

print("Estado inicial:")
fila.display()

print("\nDequeue:", fila.dequeue())
print("Dequeue:", fila.dequeue())

print("\nEstado após duas remoções:")
fila.display()

fila.enqueue(50)

print("\nApós Enqueue(50):")
fila.display()

print("\nDequeue:", fila.dequeue())

print("\nEstado final:")
fila.display()
```

A transferência de P1 para P2 preserva a propriedade FIFO porque a inversão realizada pela pilha faz com que o primeiro elemento inserido em P1 passe a ocupar o topo de P2, tornando-se o primeiro a ser removido. Por exemplo, ao executar Enqueue(1), Enqueue(2), Enqueue(3) e Enqueue(4), a pilha P1 contém os elementos na ordem [1, 2, 3, 4], sendo 4 o topo. Após a transferência, P2 passa a conter [4, 3, 2, 1], ficando 1 no topo. Assim, as chamadas sucessivas de Dequeue retornam 1, depois 2, depois 3 e finalmente 4, exatamente na mesma ordem em que foram inseridos na fila.
