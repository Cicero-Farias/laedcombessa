```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:

    def __init__(self):
        self.head = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            return

        atual = self.head
        while atual.next:
            atual = atual.next

        atual.next = novo

    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("None")
```

```python
def elemento_mais_frequente(lista):

    if lista.head is None:
        return None, 0

    frequencias = {}

    atual = lista.head

    while atual is not None:

        if atual.value in frequencias:
            frequencias[atual.value] += 1
        else:
            frequencias[atual.value] = 1

        atual = atual.next

    elemento = None
    maior_frequencia = 0

    for valor, quantidade in frequencias.items():

        if quantidade > maior_frequencia:
            maior_frequencia = quantidade
            elemento = valor

    return elemento, maior_frequencia
```

```python
lista = LinkedList()

for valor in [8, 3, 8, 5, 8, 3]:
    lista.append(valor)

print("Lista:")
lista.display()

elemento, ocorrencias = elemento_mais_frequente(lista)

print(f"\nElemento mais frequente: {elemento}")
print(f"Número de ocorrências: {ocorrencias}")
```

Tempo: O(n), pois cada nó é visitado apenas uma vez e a busca pelo maior valor no dicionário também é linear em relação ao número de elementos distintos.
