```python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None
```

```python
class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, value):

        novo = Node(value)

        if self.head is None:
            self.head = novo
            self.tail = novo
            return

        novo.prev = self.tail
        self.tail.next = novo
        self.tail = novo

    def display(self):

        atual = self.head

        while atual:
            print(atual.value, end=" <-> ")
            atual = atual.next

        print("None")
```

```python
def inverter(lista):

    atual = lista.head

    while atual is not None:

        # Troca os ponteiros
        atual.prev, atual.next = atual.next, atual.prev

        # Avança para o próximo nó (que agora está em prev)
        atual = atual.prev

    # Troca head e tail
    lista.head, lista.tail = lista.tail, lista.head
```

```python
lista = DoublyLinkedList()

for valor in [2, 4, 6, 8, 10]:
    lista.append(valor)

print("Lista original:")
lista.display()

inverter(lista)

print("\nLista invertida:")
lista.display()
```

Tempo: O(n), pois cada nó é visitado exatamente uma vez.
