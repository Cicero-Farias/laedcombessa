```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:

    def __init__(self):
        self.head = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            return

        atual = self.head

        while atual.next:
            atual = atual.next

        atual.next = novo

    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("None")
```

```python
def intersecao(lista1, lista2):

    resultado = LinkedList()

    # Armazena os elementos da segunda lista
    elementos = set()

    atual = lista2.head

    while atual is not None:
        elementos.add(atual.value)
        atual = atual.next

    # Evita inserir elementos repetidos na lista resultado
    adicionados = set()

    atual = lista1.head

    while atual is not None:

        if atual.value in elementos and atual.value not in adicionados:
            resultado.append(atual.value)
            adicionados.add(atual.value)

        atual = atual.next

    return resultado
```

```python
lista1 = LinkedList()
lista2 = LinkedList()

for valor in [3, 9, 2, 6, 4]:
    lista1.append(valor)

for valor in [4, 5, 2, 9, 3]:
    lista2.append(valor)

print("Lista 1:")
lista1.display()

print("Lista 2:")
lista2.display()

resultado = intersecao(lista1, lista2)

print("\nInterseção:")
resultado.display()
```

Tempo: O(n + m), onde n é o número de elementos da primeira lista e m o da segunda, pois cada lista é percorrida apenas uma vez.
