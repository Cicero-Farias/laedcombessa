```python
def kesimo_maior(v, k):
    if k > len(v) or k <= 0:
        return None

    v_ordenado = sorted(v, reverse=True)

    return v_ordenado[k - 1]

v = [9, 42, 21, 14, 25, 3, 19, 33, 45, 6]
print(kesimo_maior(v, 4))
```
