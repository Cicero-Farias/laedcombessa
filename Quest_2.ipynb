```python
def segundo_maior_impar(V):
    maior = None
    segundo = None
    for x in V:
        if x % 2 == 0:
            continue
        if maior is None or x > maior:
            if maior is None:
                maior = x
            else:
                segundo = maior
                maior = x
        elif x != maior and (segundo is None or x > segundo):
            segundo = x
    return segundo

V = [9, 42, 21, 14, 28, 3, 19, 32, 46, 6]
print(segundo_maior_impar(V))
```
