```python
def bubble_sort_parcial(lista, inicio, fim):
    for i in range(inicio, fim):
        for j in range(inicio, fim - (i - inicio)):
            if lista[j] > lista[j + 1]:
                temp = lista[j]
                lista[j] = lista[j + 1]
                lista[j + 1] = temp


def algoritmo(lista):
    n = len(lista)

    k = (2 * n) // 3

    bubble_sort_parcial(lista, 0, k - 1)

    bubble_sort_parcial(lista, n - k, n - 1)

    bubble_sort_parcial(lista, 0, k - 1)

    return lista


L = [9, 8, 7, 6, 5, 4, 3, 2, 1]

resultado = algoritmo(L)

print("Resultado:")
print(resultado)
```
