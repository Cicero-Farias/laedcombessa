```python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None
```

```python
class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, value):

        novo = Node(value)

        if self.head is None:
            self.head = novo
            self.tail = novo
            return

        novo.prev = self.tail
        self.tail.next = novo
        self.tail = novo

    def display(self):

        atual = self.head

        while atual:
            print(atual.value, end=" <-> ")
            atual = atual.next

        print("None")
```

```python
def busca(L, x):

    # Escolhe a sublista onde x pode estar
    for sublista in L:

        if sublista.head is None:
            continue

        if x <= sublista.tail.value:

            atual = sublista.head

            while atual:

                if atual.value == x:
                    return True

                if atual.value > x:
                    return False

                atual = atual.next

            return False

    return False
```

```python
def inserir(L, x):

    for sublista in L:

        if sublista.head is None:
            sublista.append(x)
            return

        if x <= sublista.tail.value:

            novo = Node(x)

            if x <= sublista.head.value:

                novo.next = sublista.head
                sublista.head.prev = novo
                sublista.head = novo
                return

            atual = sublista.head

            while atual.next and atual.next.value < x:
                atual = atual.next

            novo.next = atual.next
            novo.prev = atual

            if atual.next:
                atual.next.prev = novo
            else:
                sublista.tail = novo

            atual.next = novo
            return

    # Maior que todos os elementos
    L[-1].append(x)
```

```python
def remover(L, x):

    for sublista in L:

        if sublista.head is None:
            continue

        if x <= sublista.tail.value:

            atual = sublista.head

            while atual:

                if atual.value == x:

                    if atual.prev:
                        atual.prev.next = atual.next
                    else:
                        sublista.head = atual.next

                    if atual.next:
                        atual.next.prev = atual.prev
                    else:
                        sublista.tail = atual.prev

                    return True

                if atual.value > x:
                    return False

                atual = atual.next

            return False

    return False
```

```python
L = []

s1 = DoublyLinkedList()
for x in [2, 5, 9]:
    s1.append(x)

s2 = DoublyLinkedList()
for x in [15, 18, 19]:
    s2.append(x)

s3 = DoublyLinkedList()
for x in [31, 40, 49]:
    s3.append(x)

L.append(s1)
L.append(s2)
L.append(s3)

print("Sublistas:")

for i, s in enumerate(L):
    print(f"Sublista {i+1}: ", end="")
    s.display()

print("\nBusca 18:", busca(L, 18))
print("Busca 20:", busca(L, 20))

print("\nInserindo 17")
inserir(L, 17)

for i, s in enumerate(L):
    print(f"Sublista {i+1}: ", end="")
    s.display()

print("\nRemovendo 18")
remover(L, 18)
x
for i, s in enumerate(L):
    print(f"Sublista {i+1}: ", end="")
    s.display()
```
