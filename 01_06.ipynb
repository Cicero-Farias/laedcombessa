```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class Stack:

    def __init__(self):
        self.top = None

    # Insere um elemento no topo
    def push(self, value):
        novo = Node(value)
        novo.next = self.top
        self.top = novo

    # Remove o elemento do topo
    def pop(self):

        if self.top is None:
            print("Pilha vazia.")
            return None

        # Guarda o nó que será removido
        removido = self.top

        # Avança o topo
        self.top = self.top.next

        return removido.value

    # Mostra a pilha
    def display(self):

        atual = self.top

        print("topo -> ", end="")

        while atual is not None:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("/")
```

```python
pilha = Stack()

# Inserimos na ordem inversa para obter:
# topo -> 42 -> 17 -> 5

pilha.push(5)
pilha.push(17)
pilha.push(42)

print("Pilha inicial:")
pilha.display()
```

```python
pilha.push(99)
pilha.push(3)

print("Após Push(99) e Push(3):")
pilha.display()
```

```python
removido1 = pilha.pop()

print("Primeiro Pop:")
print("Valor removido:", removido1)
pilha.display()

removido2 = pilha.pop()

print("\nSegundo Pop:")
print("Valor removido:", removido2)
pilha.display()
```

É necessário guardar o nó que será removido em uma variável auxiliar antes de atualizar o ponteiro topo, pois esse nó contém a referência para o restante da pilha por meio do ponteiro next. Primeiro, salva-se o nó atual, depois o topo é atualizado para apontar para o próximo elemento e, somente então, o nó removido pode ser descartado. Se o nó fosse liberado antes da atualização de topo, perder-se-ia a referência para o restante da pilha, tornando impossível acessar os demais elementos. Assim, o uso da variável auxiliar garante que a estrutura da pilha seja preservada durante a operação de remoção.

As operações Push e Pop possuem tempo de execução O(1), pois nenhuma delas precisa percorrer os elementos da pilha. Na operação Push, basta criar um novo nó, fazê-lo apontar para o antigo topo e atualizar o ponteiro topo. Na operação Pop, apenas se guarda o nó do topo, atualiza-se o ponteiro topo para o próximo nó e retorna-se o valor removido. Como essas operações executam um número constante de instruções, independentemente da quantidade de elementos armazenados na pilha, ambas apresentam complexidade de tempo constante, isto é, O(1).
