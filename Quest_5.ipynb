```python
def k_repeticoes(v, k):
    contagem = {}

    for x in v:
        if x in contagem:
            contagem[x] += 1
        else:
            contagem[x] = 1

        if contagem[x] >= k:
            return True, x

    return False, None

v = [7, 1, 9, 1, 7, 3, 9, 2, 1, 6, 8, 3]

print(k_repeticoes(v, 3))
```
