```python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None
```

```python
class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, value):

        novo = Node(value)

        if self.head is None:
            self.head = novo
            self.tail = novo
            return

        self.tail.next = novo
        novo.prev = self.tail
        self.tail = novo

    def display(self):

        atual = self.head

        while atual:
            print(atual.value, end=" <-> ")
            atual = atual.next

        print("None")
```

```python
def atualizar(lista, x, y):

    # Procura o nó contendo x
    atual = lista.head

    while atual is not None and atual.value != x:
        atual = atual.next

    if atual is None:
        print("Elemento não encontrado.")
        return

    # Remove o nó da lista
    if atual.prev is not None:
        atual.prev.next = atual.next
    else:
        lista.head = atual.next

    if atual.next is not None:
        atual.next.prev = atual.prev
    else:
        lista.tail = atual.prev

    # Atualiza o valor
    atual.value = y
    atual.prev = None
    atual.next = None

    # Lista ficou vazia
    if lista.head is None:
        lista.head = atual
        lista.tail = atual
        return

    # Inserção no início
    if y <= lista.head.value:
        atual.next = lista.head
        lista.head.prev = atual
        lista.head = atual
        return

    # Inserção no final
    if y >= lista.tail.value:
        atual.prev = lista.tail
        lista.tail.next = atual
        lista.tail = atual
        return

    # Inserção no meio
    p = lista.head

    while p is not None and p.value < y:
        p = p.next

    atual.next = p
    atual.prev = p.prev

    p.prev.next = atual
    p.prev = atual
```

```python
lista = DoublyLinkedList()

for valor in [3, 5, 9, 10, 15]:
    lista.append(valor)

print("Original:")
lista.display()

atualizar(lista, 9, 12)

print("Depois:")
lista.display()
```

```python
lista = DoublyLinkedList()

for valor in [3, 5, 9, 10, 15]:
    lista.append(valor)

atualizar(lista, 10, 1)

lista.display()
```

```python
lista = DoublyLinkedList()

for valor in [3, 5, 9, 10, 15]:
    lista.append(valor)

atualizar(lista, 5, 20)

lista.display()
```

Tempo: O(n), pois há uma busca pelo elemento e outra busca pela posição correta de inserção.
