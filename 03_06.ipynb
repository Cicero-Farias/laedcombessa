```python
class Stack:

    def __init__(self):
        self.items = []

    def push(self, value):
        self.items.append(value)

    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0

    def display(self):
        print(self.items)
```

```python
class QueueWithStacks:

    def __init__(self):
        self.pilha_entrada = Stack()
        self.pilha_saida = Stack()

    # Enfileirar
    def enqueue(self, value):
        self.pilha_entrada.push(value)

    # Desenfileirar
    def dequeue(self):

        if self.pilha_saida.is_empty():

            while not self.pilha_entrada.is_empty():
                self.pilha_saida.push(
                    self.pilha_entrada.pop()
                )

        if self.pilha_saida.is_empty():
            return None

        return self.pilha_saida.pop()

    def display(self):
        print("Pilha de entrada:", self.pilha_entrada.items)
        print("Pilha de saída :", self.pilha_saida.items)
```

```python
fila = QueueWithStacks()

fila.enqueue(10)
fila.enqueue(20)
fila.enqueue(30)
fila.enqueue(40)

print("Estado inicial:")
fila.display()

print("\nRemovido:", fila.dequeue())
print("Removido:", fila.dequeue())

print("\nEstado após duas remoções:")
fila.display()

fila.enqueue(50)

print("\nApós inserir 50:")
fila.display()

print("\nRemovido:", fila.dequeue())

print("\nEstado final:")
fila.display()
```

A fila mantém o comportamento FIFO (First In, First Out) utilizando duas pilhas. Todos os novos elementos são inseridos na pilha de entrada. Quando é necessária uma remoção e a pilha de saída está vazia, todos os elementos da pilha de entrada são transferidos para a pilha de saída, invertendo sua ordem. Dessa forma, o primeiro elemento inserido torna-se o topo da pilha de saída e é removido primeiro, preservando exatamente o comportamento de uma fila.

A operação Enqueue consiste apenas em inserir um elemento na pilha de entrada, possuindo complexidade O(1). A operação Dequeue também é O(1) na maioria das chamadas, mas, quando a pilha de saída está vazia, é necessário transferir todos os elementos da pilha de entrada para a pilha de saída, o que custa O(n) no pior caso. Entretanto, como cada elemento é transferido no máximo uma vez entre as pilhas antes de ser removido, o custo amortizado da operação Dequeue é O(1).
