```python
def maior_impar(V):
    maior = None
    for x in V:
        if x % 2 != 0:
            if maior is None or x > maior:
                maior = x
    return maior

V = [9, 42, 21, 14, 28, 3, 19, 32, 46, 6]
print(maior_impar(V))
```
