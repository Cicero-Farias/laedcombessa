```python
def particao(lista):
    pivo = lista[0]
    menores = []
    maiores = []

    for i in range(1, len(lista)):
        if lista[i] <= pivo:
            menores.append(lista[i])
        else:
            maiores.append(lista[i])

    return menores, pivo, maiores


def bubble_sort(lista):
    n = len(lista)

    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                temp = lista[j]
                lista[j] = lista[j + 1]
                lista[j + 1] = temp

    return lista


def algoritmo(lista):
    menores, pivo, maiores = particao(lista)

    menores = bubble_sort(menores)
    maiores = bubble_sort(maiores)

    return menores + [pivo] + maiores

L = [7, 2, 9, 1, 5, 8, 3]

resultado = algoritmo(L)

print("Lista ordenada:")
print(resultado)
```
