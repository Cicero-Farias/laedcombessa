```python
def terceiro_maior(V):
    if len(V) < 3:
        raise ValueError("menos de 3 elementos")
    a, b, c = V[0], V[1], V[2]
    if a < b: a, b = b, a
    if b < c: b, c = c, b
    if a < b: a, b = b, a
    max1, max2, max3 = a, b, c
    for x in V[3:]:
        if x > max1:
            max3, max2, max1 = max2, max1, x
        elif x > max2:
            max3, max2 = max2, x
        elif x > max3:
            max3 = x
    return max3

V = [9, 42, 21, 14, 25, 3, 19, 33, 45, 6]
print(terceiro_maior(V))
```
