```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:

    def __init__(self):
        self.head = None

    def append(self, value):
        novo = Node(value)

        if self.head is None:
            self.head = novo
            return

        atual = self.head

        while atual.next:
            atual = atual.next

        atual.next = novo

    def display(self):
        atual = self.head

        while atual:
            print(atual.value, end=" -> ")
            atual = atual.next

        print("None")
```

```python
def possui_repetidos(lista):

    vistos = set()

    atual = lista.head

    while atual is not None:

        if atual.value in vistos:
            return True

        vistos.add(atual.value)
        atual = atual.next

    return False
```

```python
lista = LinkedList()

for valor in [2, 9, 7, 4, 1]:
    lista.append(valor)

print("Lista:")
lista.display()

if possui_repetidos(lista):
    print("\nA lista possui elementos repetidos.")
else:
    print("\nA lista NÃO possui elementos repetidos.")
```

```python
lista2 = LinkedList()

for valor in [2, 9, 7, 4, 2, 1]:
    lista2.append(valor)

print("Lista:")
lista2.display()

if possui_repetidos(lista2):
    print("\nA lista possui elementos repetidos.")
else:
    print("\nA lista NÃO possui elementos repetidos.")
```

Tempo: O(n), pois cada elemento da lista é visitado apenas uma vez. A busca e inserção em um set possuem custo médio O(1).
