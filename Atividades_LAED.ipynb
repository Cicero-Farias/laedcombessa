```python
def maior_impar(V):
    maior = None
    for x in V:
        if x % 2 != 0:
            if maior is None or x > maior:
                maior = x
    return maior

V = [9, 42, 21, 14, 28, 3, 19, 32, 46, 6]
print(maior_impar(V))
```

```python
def segundo_maior_impar(V):
    maior = None
    segundo = None
    for x in V:
        if x % 2 == 0:
            continue
        if maior is None or x > maior:
            if maior is None:
                maior = x
            else:
                segundo = maior
                maior = x
        elif x != maior and (segundo is None or x > segundo):
            segundo = x
    return segundo

V = [9, 42, 21, 14, 28, 3, 19, 32, 46, 6]
print(segundo_maior_impar(V))
```

```python
def busca_aproximada(V, k):
    if not V:
        return None, False

    melhor = V[0]
    for x in V:
        if x == k:
            return k, True
        if abs(x - k) < abs(melhor - k):
            melhor = x
        elif abs(x - k) == abs(melhor - k) and x < melhor:
            melhor = x
    return melhor, False

V = [9, 42, 21, 14, 25, 3, 19, 33, 45, 6]
k = 31
resultado, achou = busca_aproximada(V, k)
if achou:
    print(f"{resultado} (está lá)")
else:
    print(f"Não está lá, mas eu encontrei o {resultado}")
```

```python
from collections import Counter

def impar_impar(V):
    cont = Counter(V)
    for valor, freq in cont.items():
        if valor % 2 != 0 and freq % 2 != 0:
            return valor
    return None

V = [9, 2, 7, 7, 2, 2, 1, 7, 7, 9]
print(impar_impar(V))
```

```python
def alguem_e_dobro_nao_ordenada(V):
    seen = set()
    for x in V:
        if 2 * x in seen or (x % 2 == 0 and (x // 2) in seen):
            if 2 * x in seen:
                return (x, 2 * x)
            else:
                return (x // 2, x)
        seen.add(x)
    return None

V = [9, 42, 21, 14, 25, 3, 19, 33, 45, 6]
print(alguem_e_dobro_nao_ordenada(V))
```

```python
def alguem_e_dobro_ordenada(V):
    i, j = 0, 0
    n = len(V)
    while i < n and j < n:
        if i == j:
            j += 1
            continue
        if V[j] == 2 * V[i]:
            return (V[i], V[j])
        if V[j] < 2 * V[i]:
            j += 1
        else:
            i += 1
    return None

V_sorted = sorted([9, 42, 21, 14, 25, 3, 19, 33, 45, 6])
print(V_sorted, alguem_e_dobro_ordenada(V_sorted))
```

```python
def dois_mais_proximos(V):
    n = len(V)
    if n < 2:
        return None

    V_sorted = sorted(V)
    minimo = abs(V_sorted[1] - V_sorted[0])
    par = (V_sorted[0], V_sorted[1])

    for i in range(1, n-1):
        a, b = V_sorted[i], V_sorted[i+1]
        d = abs(b - a)
        if d < minimo:
            minimo = d
            par = (a, b)

    return par[0], par[1], minimo

V = [9, 2, 7, 7, 2, 2, 1, 7, 7, 9]
print(dois_mais_proximos(V))
```

```python
def contar_inversoes(V):
    def merge_count(left, right):
        i = j = 0
        merged = []
        inv = 0
        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                merged.append(left[i])
                i += 1
            else:
                merged.append(right[j])
                j += 1
                inv += len(left) - i
        if i < len(left):
            merged.extend(left[i:])
        if j < len(right):
            merged.extend(right[j:])
        return merged, inv

    def sort_count(arr):
        n = len(arr)
        if n <= 1:
            return arr, 0
        mid = n // 2
        left, inv_left = sort_count(arr[:mid])
        right, inv_right = sort_count(arr[mid:])
        merged, inv_merge = merge_count(left, right)
        return merged, inv_left + inv_right + inv_merge

    _, inversoes = sort_count(list(V))
    return inversoes

V = [9, 2, 7, 7, 2, 2, 1, 7, 7, 9]
print(contar_inversoes(V))
```

```python
def intersecao(U, V):
    if not U or not V:
        return []
    if len(U) <= len(V):
        menor, maior = U, V
    else:
        menor, maior = V, U

    conjunto_menor = set(menor)
    inter = set()
    for x in maior:
        if x in conjunto_menor:
            inter.add(x)

    return sorted(inter)

U = [9, 2, 7, 7, 2, 2, 1, 7, 7, 9]
V = [2, 15, 19, 12, 33, 9, 17, 41, 54, 8]
print(intersecao(U, V))
```
